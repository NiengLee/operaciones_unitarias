<h1 style="text-align:center;">Capítulo 13 - Flujo en medio poroso.</h1>
<p style="text-align:center;">En esta sección se explica como se separan sólidos suspendidos en un fluido y qué criterio se usa para elegir el método de separación. <p>

Un medio poroso es un sólido que contiene espacios vacíos interconectados por donde puede circular un fluido. Puede representarse como un material continuo con poros internos (medio consolidado) o como un conjunto de partículas empacadas con huecos entre ellas (medio no consolidado).

La propiedad fundamental es la porosidad, que mide la fracción del volumen ocupada por vacíos y determina cuánto espacio tiene el fluido para moverse. En este tipo de medios se distinguen dos velocidades: 
- la velocidad superficial, calculada sobre el área total
- la velocidad intersticial, que es la velocidad real del fluido dentro de los poros.


# Filtración:

Recomendable cuando las partículas sólidas son muy pequeñas, no sedimentan de forma eficiente, por lo que se usa un método de “barrera”: el fluido atraviesa un medio filtrante que retiene los sólidos.

Ese medio filtrante (como tela o malla) actúa como soporte, pero no necesariamente atrapa directamente las partículas por tamaño. En muchos casos, los poros del filtro son incluso más grandes que las partículas, por lo que estas penetran parcialmente dentro del material antes de quedar atrapadas.

Con el tiempo, las partículas retenidas se acumulan en la superficie del filtro formando una capa sólida llamada “cake” (torta de filtración). Esta capa es la que realmente realiza la mayor parte de la separación, ya que funciona como un nuevo medio poroso que controla el paso del fluido.

El comportamiento del proceso depende principalmente de cómo fluye el líquido a través de ese cake, es decir, de la relación entre presión aplicada y caudal que atraviesa la estructura porosa formada por los sólidos acumulados.

Recordando la ecuación de "Hagen-Poiseuille" (Tomado del libro de fenomenos de transporte del autor Bird):



$$
\begin{aligned}
Q = \frac{\pi R^4 \Delta P}{8 \mu L} \\[10pt]
\end{aligned}
$$

Se reduce el termino $\frac{R^2}{8}$ a una constante k de permeabilidad, que tambien se puede calcular de la siguiente manera:

Capítulo 8 : flujo no confinado o flujo alrededor de objetos, 8.2.2 Modelos de Comportamiento de Filtros de Torta , página 173



$$
\begin{aligned}
k = \frac{d_p² Re}{2 F_f} \\[10pt]
\end{aligned}
$$

luego reordenamos, y tenemos la ley de Darcy para medios porosos:



$$
\begin{aligned}
\vec{v}_0 = \frac{Q}{A} &= \frac{1}{A}\frac{dV}{dt} = \frac{k (-\Delta P)}{\mu L} \\[10pt]
Q &= \frac{-kA \Delta P}{\mu L} \\[10pt]
\frac{-\Delta P}{L} &= \frac{\mu Q}{-kA} \\[10pt]
\end{aligned}
$$

Como vemos, el filtro se compone tanto de torta adherida en el medio poroso como medio poroso, así:

$$
\begin{aligned}
P_1 - P_2 &= \left(\frac{L}{k}\right)_{torta} \frac{Q\mu}{A} \\[10pt]
P_2 - P_3 &= \left(\frac{L}{k}\right)_{filtro} \frac{Q\mu}{A} \\[10pt]
P_1 - P_3 &= \frac{Q\mu}{A} \left[\left(\frac{L}{k}\right)_{torta} + \left(\frac{L}{k}\right)_{filtro}\right]\\[10pt]
\end{aligned}
$$

Analizamos el termino L de la sección de la torta:

$$
\begin{aligned}
L_{torta} &= \frac{\~V_{torta}}{A} \\[10pt]
\epsilon = \frac{\text{Volumen total} - \text{Volumen solidos}}{\text{Volumen total}} &= 1 - \frac{A_{solida}}{A_{total}} = \frac{A_{vacio}}{A_{total}} \\[10pt]
L_{torta} &= \frac{\~V_{solidos}}{A(1-\epsilon)} = \frac{M_{solidos}}{A\rho_{solidos}(1-\epsilon)} \\[10pt]
\end{aligned}
$$

A la ultima ecuación, amplificamos por $\~V$ en numerador y denominador:
$$
\begin{aligned}
L_{torta} &= \frac{M_{solidos}}{A\rho_{solidos}(1-\epsilon)} \frac{\~V}{\~V} \\[10pt]
W &= \frac{M_{solidos}}{\~V\rho_{solidos}(1-\epsilon)} \\[10pt]
L_{torta} &= W \frac{\~V}{A} \\[10pt]
\end{aligned}
$$

y la densidad de la torta se determina mediante : $\rho_{torta} = (1 - \epsilon)\rho_{solido} + \epsilon \rho_{liquido}$. Tenemos finalmente:

$$
\begin{aligned}
P_1 - P_3 &= \frac{Q\mu}{A} \left[\left(\frac{L}{k}\right)_{torta} + \left(\frac{L}{k}\right)_{filtro}\right] \\[10pt]
\left(\frac{L}{k}\right)_{filtro} &= a \\[10pt]
\frac{Q}{A} &= \frac{P_1 - P_3}{\mu \left[ \frac{W \~V}{Ak} + a \right]} \\[10pt]
\end{aligned}
$$

finalmente tenemos una expresión que relaciona el flujo despues del flujo vs el delta de presión:

$$
\begin{aligned}
\frac{d\~V}{Adt} &= \frac{P_1 - P_3}{\mu \left[ \frac{W \~V}{Ak} + a \right]} \\[10pt]
\end{aligned}
$$

# 1) Operaciónes a presión constante

Desarrollando la ecuación diferencial anterior a cambio de presión constante en el tiempo:

$$
\begin{aligned}
(-\Delta P) t &= C_1\left(\frac{\~V}{A}\right)^2 + C_2\left(\frac{\~V}{A}\right) \\[10pt]
C_1 &= \frac{\mu W}{2K}\\[10pt]
C_2 &= \mu a\\[10pt]
\end{aligned}
$$

Y reordenando la expresión anterior:

$$
\begin{aligned}
\left(\frac{-\Delta P}{\frac{\~V}{A}}\right)t  &= C_1\left(\frac{\~V}{A}\right) + C_2 \\[10pt]
\end{aligned}
$$

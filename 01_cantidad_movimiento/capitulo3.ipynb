<h1 style="text-align:center;">Capítulo 03 - Flujo de fluidos compresibles.</h1>
<p style="text-align:center;">En esta sección analizaremos como se abordan situaciones de flujo para fluidos compresibles, y como cambian sus propiedades y como se dan sus comportamientos, ya que al cambiar la presión o temperatura de un fluido compresible, cambiara tambien la densidad del mismo.<p>

# 0. Definiciones:

### Upstream (aguas arriba):

- región antes del flujo
- lado de entrada
- donde la presión normalmente es más alta

### Downstream (aguas abajo):

- región hacia donde se mueve el fluido
- lado de salida
- donde la presión normalmente es más baja

# 1. Diferencias clave con flujo incompresible

La diferencia central respecto al flujo incompresible es que en gases y vapores la densidad cambia durante el flujo.

## En líquidos normalmente (incompresible):

- la densidad se considera constante
- los cambios de presión afectan principalmente velocidad y pérdidas
- la caída de presión produce principalmente disipación mecánica
- la velocidad cambia por cambios geométricos (Area de flujo)
- la temperatura normalmente tiene poca importancia
- la gravedad puede ser relevante

## En gases (compresible):

- la presión modifica la densidad
- la temperatura modifica la densidad
- la velocidad modifica temperatura y presión
- todos los efectos quedan acoplados
- la caída de presión cambia la densidad
- la velocidad aumenta naturalmente al disminuir densidad
- aparecen cambios importantes de temperatura
- aparecen efectos acústicos
- la velocidad del sonido se vuelve una referencia física crítica
- los efectos gravitacionales suelen despreciarse por la baja densidad del gas

los gases convierten fácilmente energía mecánica en cambios térmicos debido a baja densidad y la baja capacidad calorífica volumétrica

# 3. Relación entre presión, densidad y velocidad

Cuando un gas pierde presión:

- la densidad disminuye
- el gas debe acelerar para conservar el flujo másico

Eso significa:

- la velocidad aumenta naturalmente a lo largo de una tubería
- la energía cinética crece
- la temperatura puede disminuir

Esto NO ocurre de forma importante en líquidos.

# 4. Existencia de un límite de velocidad

La velocidad del gas NO puede crecer indefinidamente, existe un límite físico: la velocidad del sonido en el medio

Ese límite define:

- el flujo sónico
- el choking
- la máxima capacidad de transporte de una línea

# 5. Velocidad del sonido

el sonido se presenta como:

- una onda de presión: una perturbación que viaja dentro del fluido transportando cambios de presión
- una perturbación compresiva propagándose en el gas

Conceptualmente:

- los cambios de presión viajan a velocidad finita
- esa velocidad es la velocidad del sonido

la presión NO se transmite instantáneamente en gases


# 6. Número de Mach

Representa la relación entre velocidad del fluido y velocidad del sonido

Clasificacion el régimen del flujo:

- Subsónico: el fluido viaja más lento que el sonido
- Sónico: el fluido viaja exactamente a velocidad del sonido
- Supersónico: el fluido supera la velocidad del sonido

si Mach < 0.3 :normalmente el flujo puede tratarse como incompresible!


# 7. Choked flow (flujo estrangulado)

El choking ocurre cuandoel gas alcanza velocidad sónica en algún punto del sistema

Consecuencias:

- el flujo másico alcanza un máximo
- seguir reduciendo presión downstream ya no aumenta el caudal
- el sistema deja de responder hidráulicamente a reducciones adicionales de presión externa
- indica la capacidad máxima de transporte de una línea

la información de presión no puede viajar upstream si el gas ya viaja a velocidad del sonido

# 8. Flujo isotérmico

flujo donde la temperatura constante ($\Delta T = 0$)

Ocurre en:

- líneas largas
- suficiente tiempo de residencia
- intercambio térmico apreciable con alrededores
- tuberías no aisladas

Características:

- el sistema intercambia calor continuamente
- el calor compensa enfriamientos asociados a expansión
- la temperatura permanece aproximadamente constante

Aspectos clave:

- el equilibrio térmico con el entorno es el mecanismo físico dominante
- es una buena aproximación para ductos largos

# 9.Flujo adiabático

No hay transferencia neta de calor ($Q = 0$)

Ocurre en:

- líneas aisladas
- tiempos de residencia cortos
- altas velocidades
- válvulas
- orificios
- toberas

Características:

- la temperatura cambia internamente
- la energía térmica se transforma en energía cinética
- aparecen enfriamientos importantes durante expansión

### locally isentropic approach (suposicion de isoentropia local)

porque aunque existe fricción localmente el comportamiento puede aproximarse como isoentrópico

# 10. Flujo isoentrópico

Condición entropía aproximadamente constante

Ocurre en:

- trayectos extremadamente cortos
- toberas
- válvulas
- orificios
- regiones con pérdidas muy pequeñas

Características:

- fricción despreciable
- transferencia de calor despreciable
- comportamiento altamente reversible localmente

Enfoques :

- “localmente isoentrópico” porque: el fenómeno ocurre en regiones pequeñas

# 11. Régimen politrópico

Es un  comportamiento intermedio entre isotérmico y adiabátic y es importante analizarporque

- muchos sistemas reales NO son perfectamente isotérmicos
- tampoco perfectamente adiabáticos

Entonces el comportamiento real cae entre ambos extremos

# 12. factor de expansión

corrige Bernoulli incompresible para incluir compresibilidad y representa cuánto difiere el flujo real compresible respecto al cálculo incompresible

Importancia: permite usar formulaciones tipo Bernoulli incluyendo expansión del gas

# 13. Ecuación de Weymouth

Es una  aproximación simplificada para flujo compresible isotérmico, desprecia contribuciones energéticas asociadas a aceleración del gas

Interpretación:

el sistema se aproxima al comportamiento incompresible usando densidad promedio

# 14. Tobera (nozzle)

Tipos:

- convergente
- divergente
- convergente-divergente

Importancia:

- controlar aceleración del gas
- inducir transición subsónico/supersónico
- evitar ondas de choque destructivas

Aplicaciones mencionadas:

- turbinas
- oxicorte
- combustores
- sand blasting
- aeronáutica

# 15. Comportamiento subsónico vs supersónico

- En régimen subsónico reducir área acelera el flujo: Comportamiento parecido al flujo incompresible.
- En régimen supersónico: reducir área desacelera el flujo; aumentar área acelera el flujo

Contrario a la intuición hidráulica clásica.

Por eso: las toberas convergente-divergente son necesarias para alcanzar velocidades supersónicas

# 16. Fanno line

Describe flujo adiabático con fricción en ductos de área constante

Características:

- la fricción hace evolucionar el flujo hacia condición sónica
- tanto flujos subsónicos como supersónicos tienden hacia Mach 1

Resultado:

- el flujo sónico actúa como estado límite

# 17. Rol de la fricción

- En líquidos: la fricción principalmente destruye energía mecánica
- En gases:
> - la fricción también modifica:
> - temperatura
> - densidad
> - velocidad
> - régimen de Mach

Además puede inducir choking

# 18. Rol del Reynolds

- el factor de fricción depende de Reynolds
- Reynolds depende del flujo
- el problema queda acoplado e implícito

Consecuencia:

normalmente se requieren iteraciones

# 19.Importancia industrial

Importancia en:

- líneas de vapor
- aire comprimido
- sistemas neumáticos
- válvulas
- impulsores
- sistemas de combustión
- expansión de gases
- equipos de proceso

### Tres ideas fundamentales:

- La compresibilidad aparece cuando presión y temperatura modifican apreciablemente la densidad.
- El número de Mach controla el comportamiento físico del flujo.
- El choking establece el límite máximo de transporte de masa en un conducto compresible.

# Ecuaciones de flujo de fluidos compresibles

### Expresiones comunes:

$$
\begin{aligned}
\checkmark \text{Número Mach} &\rightarrow Ma = \frac{\vec{v}}{c} \\[10pt]
\checkmark \text{Flux másico} &\rightarrow G = \frac{\vec{\dot m}}{A} = \rho \vec{v} \\[10pt]
\checkmark \text{Velocidad de flujo} &\rightarrow \vec{v} = \sqrt{\frac{2k}{k-1}\frac{P_1}{\rho_1}\left[ 1 - \left(\frac{P_2}{P_1}\right)^{\frac{k-1}{k}}\right]} \\[10pt] 
\checkmark \text{Velocidad del sonido en el fluido} &\rightarrow c = \sqrt{\frac{kRT}{PM}} \\[10pt]
\end{aligned}
$$

- c : velocidad del sonido en el fluido

### Régimen Isotérmico:

$$
\begin{aligned}
\checkmark \text{Flux másico} &\rightarrow G = \sqrt{\frac{\rho_1 P_1 \left[1 - \left(\frac{P_2}{P_1}\right)^2 \right]}{\left[\frac{L}{D}f - ln\left(\frac{P_2}{P_1}\right)^2 \right]}} \\[10pt]
\checkmark \text{Calor transferido} &\rightarrow q = \frac{G^2}{2\~\rho^2} \left[\frac{\rho_1}{G^2P_1}\left[P_2^2 - P_1^2\right] + \frac{L}{D}f\right] \\[10pt]
\checkmark \text{Flujo de obstrucción} &\rightarrow G^* = P_2^* \sqrt{\frac{\rho_1}{P_1}} \\[10pt]
&\rightarrow \Sigma_i K_f = \left(\frac{P_1}{P_2^*}\right)^2 - 2ln\left(\frac{P_1}{P_2^*}\right) - 1 \\[10pt]
\end{aligned}
$$

- $\~\rho$ : densidad promedio
- $*$ : "Choked flow", es el flujo de obstrucción (estrangulado, sonido crítico), condición crítica, sónica, donde se alcanza $Ma = 1$, es decir, la velocidad del gas es igual a la velocidad del sonido en el gas
- $G^*$ : máximo flujo másico posible antes del choking
- $P^*$ : presión crítica de salida donde ocurre choking.
- Para realizar iteraciones y encontrar el flujo de obstrucción es necesario tambien contar con la ecuacion de perdidas propuesta por shacham o semejantes, y el factor adimensional de $Re = \frac{DG}{\mu}$

### Régimen adiabatico - Localmente isentrópico

$$
\begin{aligned}
\checkmark \text{Flux másico} &\rightarrow G² = 2\rho_1 P_1 \left(\frac{1}{\sigma}\right) \frac{\left[1 - \left(\frac{P_2}{P_1}\right)^{\sigma} \right]}{\left[\frac{L}{D}f - \frac{2}{k}\mathrm{ln}\left(\frac{P_2}{P_1}\right) \right]} \\[10pt]
\checkmark \text{Relaciones térmicas} &\rightarrow \frac{2Cp\rho_1^2}{(\theta + 1)G^2T_1^{\theta}}\left[T_2^{\theta +  1} - T_1^{\theta +  1}\right] - \frac{2\rho_1}{\sigma G^2P_1^{\frac{1}{k}}}\left[P_2^{\sigma}-P_1^{\sigma}\right] - \frac{L}{D}f = 0 \\[10pt]
\checkmark \text{Flujo de obstrucción} &\rightarrow G^* = \sqrt{P_1 \rho_1 k \left(\frac{P_2^*}{P_1}\right)^{\sigma}} \\[10pt]

&\rightarrow \Sigma_i K_f = \frac{2}{k+1} \left[\left(\frac{P_1}{P_2^*}\right)^{\sigma} - 1 \right] - \frac{2}{k}\mathrm{ln}\left(\frac{P_1}{P_2^*}\right) \\[10pt]
\end{aligned}
$$

- $\theta = \frac{2}{k-1}$
- $\sigma = \frac{k+1}{k}$
- $k = \frac{Cp}{Cv}$

### Régimen No-isotérmico - Iso-entrópico

$$
\begin{aligned}
\checkmark \text{Flux másico} &\rightarrow G² = \rho^2 \vec{v}^2 = \left(\frac{2k}{k-1}\right) P_1 \rho \left[1 - \left(\frac{P_2}{P_1}\right)^{\frac{k-1}{k}}\right] \\[10pt]
\checkmark \text{Ecuación de energía} &\rightarrow \vec{v_2} = \vec{v_1} \\[10pt]
\checkmark \text{Condiciones críticas en garganta} &\rightarrow \dot{m}_{Ct} = A_G\sqrt{P_1 \rho_1 k \left(\frac{2}{k+1}\right)^{\frac{k+1}{k-1}}} \\[10pt]
&\rightarrow P_{Ct} = P_1\left(\frac{2}{k+1}\right)^{\frac{k}{k-1}} \\[10pt]
&\rightarrow \vec{v}_{Ct} = \sqrt{\frac{2k}{k+1}\frac{P_1}{\rho_1}} \\[10pt]
\end{aligned}
$$

- $\vec{v}_{Ct}$ : es la velocidad crítica, velocidad sónica crítica, o sea la velocidad del fluido cuando ocurre choking, cuando la velocidad de fluido es igual a la velocidad del sonido en el fluido. ($Ma=1$ y $\vec{v}_{Ct} = c$)

## ACLARACIÓN :
Un flujo adiabático es un flujo donde no hay transferencia de calor con el entorno, pero sí pueden existir fricción e irreversibilidades, por lo que la entropía puede aumentar. En cambio, un flujo isoentrópico es un caso ideal de flujo adiabático donde además no hay generación de entropía, es decir, el proceso es reversible. Por eso, todo flujo isoentrópico es adiabático, pero no todo flujo adiabático es isoentrópico.

### Factor de expansión:

Es una corrección con un factor de expansión a la ecuación de Bernoulli de fluidos incompresibles a fluidos compresibles.

$$
\begin{aligned}
G &= Y \sqrt{\frac{2\rho_1 \Delta P}{\Sigma K_f}} = Y\sqrt{\rho_1 P_1} \sqrt{\frac{2\left(1-\frac{P_2}{P_1}\right)}{\Sigma K_f}} \\[10pt]
Y &= \frac{G_{adiabatico}}{G_{incompresible}} \\[10pt]
\end{aligned}
$$
